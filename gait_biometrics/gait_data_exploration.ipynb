### This notebook contains:

* File inspection
* Column inspection
* Row counts
* Timestamp analysis
* Sampling-rate estimation
* Accelerometer/gyroscope alignment analysis


```python
# Connect to the Drive dataset folder and list all CSV files
from pathlib import Path
import pandas as pd

DATA_DIR = Path("/content/drive/MyDrive/sample_data")

files = sorted(DATA_DIR.glob("*.csv"))

print(f"Number of CSV files found: {len(files)}")

for file in files:
    print(file.name)
```

```python
# Inspect columns, rows, and sample values
for file in files:
    df = pd.read_csv(file)

    print("=" * 80)
    print(f"File: {file.name}")
    print(f"Rows: {len(df):,}")
    print(f"Columns: {list(df.columns)}")
    print("\nData types:")
    print(df.dtypes)
    print("\nFirst 3 rows:")
    display(df.head(3))
```

```python
# Identify and list accelerometer and gyroscope files
accelerometer_files = []
gyroscope_files = []

for file in files:
    filename = file.name.lower()

    if "accelerometer" in filename:
        accelerometer_files.append(file)
    elif "gyroscope" in filename:
        gyroscope_files.append(file)

print(f"Accelerometer files: {len(accelerometer_files)}")
print(f"Gyroscope files: {len(gyroscope_files)}")

print("\nAccelerometer files:")
for file in accelerometer_files:
    print(file.name)

print("\nGyroscope files:")
for file in gyroscope_files:
    print(file.name)
```

```python
# Compare row counts
print("Accelerometer row counts:")
for file in sorted(accelerometer_files):
    df = pd.read_csv(file)
    print(f"{file.name}: {len(df):,} rows")

print("\nGyroscope row counts:")
for file in sorted(gyroscope_files):
    df = pd.read_csv(file)
    print(f"{file.name}: {len(df):,} rows")
```

```python
# Inspect user 2
import re
import numpy as np
import pandas as pd

# Select one user to inspect in detail
user_id = "2"

acc_file = next(
    file for file in accelerometer_files
    if file.name.startswith(f"{user_id}_")
)

gyro_file = next(
    file for file in gyroscope_files
    if file.name.startswith(f"{user_id}_")
)

acc_df = pd.read_csv(acc_file)
gyro_df = pd.read_csv(gyro_file)

print("Accelerometer columns:")
print(acc_df.columns.tolist())

print("\nGyroscope columns:")
print(gyro_df.columns.tolist())

print("\nAccelerometer data types:")
print(acc_df.dtypes)

print("\nGyroscope data types:")
print(gyro_df.dtypes)

print("\nFirst 5 accelerometer rows:")
display(acc_df.head())

print("\nFirst 5 gyroscope rows:")
display(gyro_df.head())
```

```python
# Display timestamp examples and recording start/end times
print("Accelerometer time values:")
display(acc_df["time"].head(10))

print("\nGyroscope time values:")
display(gyro_df["time"].head(10))

print("\nLast accelerometer time:")
print(acc_df["time"].tail(1).iloc[0])

print("\nLast gyroscope time:")
print(gyro_df["time"].tail(1).iloc[0])
```

```python
# Calculate timestamp differences and estimate the sensor frequency as approximately 100 Hz
def inspect_timestamps(df, name):
    # Convert timestamps into elapsed time
    time_values = pd.to_datetime(df["time"].astype(str))

    time_differences = time_values.diff().dt.total_seconds().dropna()

    print(f"\n{name}")
    print("-" * 50)
    print(f"Timestamp values: {len(time_values):,}")
    print(f"Missing timestamps: {time_values.isna().sum()}")
    print(f"Time is monotonic: {time_values.is_monotonic_increasing}")
    print(f"Total duration: {time_values.iloc[-1] - time_values.iloc[0]}")
    print(f"Minimum time difference: {time_differences.min():.6f} seconds")
    print(f"Median time difference: {time_differences.median():.6f} seconds")
    print(f"Maximum time difference: {time_differences.max():.6f} seconds")

    estimated_frequency = 1 / time_differences.median()

    print(f"Estimated sampling frequency: {estimated_frequency:.2f} Hz")


inspect_timestamps(acc_df, "Accelerometer")
inspect_timestamps(gyro_df, "Gyroscope")
```

```python
# Match each user’s accelerometer file with their gyroscope file and compare row counts
def get_user_id(file):
    return re.match(r"^(\d+)_", file.name).group(1)


acc_by_user = {get_user_id(file): file for file in accelerometer_files}
gyro_by_user = {get_user_id(file): file for file in gyroscope_files}

print("User pairing summary")
print("-" * 70)

for user_id in sorted(acc_by_user, key=int):
    acc_data = pd.read_csv(acc_by_user[user_id])
    gyro_data = pd.read_csv(gyro_by_user[user_id])

    print(
        f"User {user_id:>3}: "
        f"Accelerometer = {len(acc_data):>6,} rows | "
        f"Gyroscope = {len(gyro_data):>6,} rows | "
        f"Difference = {len(acc_data) - len(gyro_data):>2}"
    )
```

```python
# Check duplicate timestamps, large gaps, and exact timestamp overlap for user 2
def analyze_pair(user_id):
    acc_file = acc_by_user[user_id]
    gyro_file = gyro_by_user[user_id]

    acc = pd.read_csv(acc_file)
    gyro = pd.read_csv(gyro_file)

    # Convert timestamp strings into datetime values
    acc["time"] = pd.to_datetime(acc["time"])
    gyro["time"] = pd.to_datetime(gyro["time"])

    # Calculate time differences in seconds
    acc_diffs = acc["time"].diff().dt.total_seconds().dropna()
    gyro_diffs = gyro["time"].diff().dt.total_seconds().dropna()

    print(f"USER {user_id}")
    print("=" * 70)

    print(f"Accelerometer rows: {len(acc):,}")
    print(f"Gyroscope rows:     {len(gyro):,}")

    print("\nAccelerometer timing:")
    print(f"  Duplicate timestamps: {(acc['time'].diff() == pd.Timedelta(0)).sum()}")
    print(f"  Gaps greater than 20 ms: {(acc_diffs > 0.020).sum()}")
    print(f"  Largest gap: {acc_diffs.max():.3f} seconds")

    print("\nGyroscope timing:")
    print(f"  Duplicate timestamps: {(gyro['time'].diff() == pd.Timedelta(0)).sum()}")
    print(f"  Gaps greater than 20 ms: {(gyro_diffs > 0.020).sum()}")
    print(f"  Largest gap: {gyro_diffs.max():.3f} seconds")

    # Compare exact timestamp sequences
    common_times = set(acc["time"]).intersection(set(gyro["time"]))

    print("\nTimestamp alignment:")
    print(f"  Common timestamps: {len(common_times):,}")
    print(f"  Accelerometer-only timestamps: {len(set(acc['time']) - set(gyro['time']))}")
    print(f"  Gyroscope-only timestamps: {len(set(gyro['time']) - set(acc['time']))}")

    print("\nFirst 10 time differences:")
    print("Accelerometer:", acc_diffs.head(10).tolist())
    print("Gyroscope:    ", gyro_diffs.head(10).tolist())


analyze_pair("2")
```

```python
# Perform the same timestamp analysis for users 24, 49, 70, and 78
for user_id in ["24", "49", "70", "78"]:
    analyze_pair(user_id)
```

```python
# Summarize duplicates and large gaps for every user
print("Timestamp quality across all users")
print("=" * 90)

for user_id in sorted(acc_by_user, key=int):
    acc = pd.read_csv(acc_by_user[user_id])
    gyro = pd.read_csv(gyro_by_user[user_id])

    acc["time"] = pd.to_datetime(acc["time"])
    gyro["time"] = pd.to_datetime(gyro["time"])

    acc_diffs = acc["time"].diff().dt.total_seconds().dropna()
    gyro_diffs = gyro["time"].diff().dt.total_seconds().dropna()

    print(
        f"User {user_id:>3} | "
        f"Acc duplicate: {(acc_diffs == 0).sum():>3} | "
        f"Gyro duplicate: {(gyro_diffs == 0).sum():>3} | "
        f"Acc gaps >20ms: {(acc_diffs > 0.020).sum():>3} | "
        f"Gyro gaps >20ms: {(gyro_diffs > 0.020).sum():>3}"
    )
```

```python
# Confirm that EID correctly aligns accelerometer and gyroscope rows for user 2
def analyze_eid_alignment(user_id):
    acc = pd.read_csv(acc_by_user[user_id])
    gyro = pd.read_csv(gyro_by_user[user_id])

    acc["time"] = pd.to_datetime(acc["time"])
    gyro["time"] = pd.to_datetime(gyro["time"])

    merged = pd.merge(
        acc,
        gyro,
        on="EID",
        how="inner",
        suffixes=("_acc", "_gyro")
    )

    # Difference between accelerometer and gyroscope timestamps
    timestamp_difference = (
        merged["time_acc"] - merged["time_gyro"]
    ).dt.total_seconds().abs()

    print(f"USER {user_id}")
    print("=" * 70)
    print(f"Accelerometer rows: {len(acc):,}")
    print(f"Gyroscope rows:     {len(gyro):,}")
    print(f"Rows matched by EID: {len(merged):,}")

    print("\nEID information:")
    print(f"Accelerometer EIDs unique: {acc['EID'].is_unique}")
    print(f"Gyroscope EIDs unique:     {gyro['EID'].is_unique}")
    print(f"Accelerometer EIDs monotonic: {acc['EID'].is_monotonic_increasing}")
    print(f"Gyroscope EIDs monotonic:     {gyro['EID'].is_monotonic_increasing}")

    print("\nTimestamp differences after matching by EID:")
    print(f"Maximum difference: {timestamp_difference.max():.6f} seconds")
    print(f"Mean difference:    {timestamp_difference.mean():.6f} seconds")
    print(f"Median difference:  {timestamp_difference.median():.6f} seconds")
    print(f"Differences > 20 ms: {(timestamp_difference > 0.020).sum()}")

    print("\nFirst 10 matched rows:")
    display(
        merged[
            [
                "EID",
                "time_acc",
                "time_gyro",
                "Xvalue_acc",
                "Xvalue_gyro"
            ]
        ].head(10)
    )

    return merged


merged_user_2 = analyze_eid_alignment("2")
```

```python
# Confirm that nearly all rows match by EID for every user
print("EID alignment across all users")
print("=" * 90)

for user_id in sorted(acc_by_user, key=int):
    acc = pd.read_csv(acc_by_user[user_id])
    gyro = pd.read_csv(gyro_by_user[user_id])

    merged = pd.merge(acc, gyro, on="EID", how="inner")

    print(
        f"User {user_id:>3} | "
        f"Acc rows: {len(acc):>6,} | "
        f"Gyro rows: {len(gyro):>6,} | "
        f"Matched EIDs: {len(merged):>6,} | "
        f"Missing rows: {max(len(acc), len(gyro)) - len(merged):>2}"
    )
```
